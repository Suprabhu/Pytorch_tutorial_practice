```python
#importing libraries
import torch
import torch.nn as nn
import torch.optim as optim

# torch - pytorch core functionality
# nn - components for building neural networks
# optim - tools for training these networks
```

```python
# distance in miles
distances = torch.tensor([[1.0],[2.0],[3.0],[4.0]], dtype=torch.float32)
# delivery times in minutes
time = torch.tensor([[6.96],[12.11],[16.77],[22.21]], dtype=torch.float32)

#think of tensors as containers that store and organise data in a way that models understand
# [[1.0],[2.0],[3.0],[3.0],[4.0]] -- outer [] represent a subset of entire dataset -- collection of individual data points known as a batch
                                #-- Each inner [] represent a sample in the batch
```

```python
print(distances)
```

```python
print(time)
```

```python
#Define the model
model = nn.Sequential(nn.Linear(1,1))
#Sequential : Pytorch's streamline version: It is a conatiner that passes data through layers in order
#nn.Linear : Using only one layer, technically a linear layer
#(1,1): first value 1 means takes in one input, second value 1 means produces only one output
#applies a linear transformation to the input
```

```python
#define loss function and optimiser
loss_function = nn.MSELoss()
#Mean Square error loss: will measure how wrong or right the predictions are
optimizer = optim.SGD(model.parameters(), lr=0.01)
#optim.SGD SGD-Stochastic gradient descent: It is an algo. that figures out which direction the weight and bias to reduce that error
#model.parameters() - how you access those values in pytorch which in this case is weight and bias
# lr - learning rate: Smaller values mean you adjust your weight and bias parameters by smaller amounts.Larger values mean bigger adjustments.
```

```python
# Training loop: Actual learning happens
#Each pass through the full training data is called an epoch

for epoch in range(500):
  #in each epoch the model will:
  # 0. reset the optimizer
  optimizer.zero_grad()
  # 1.Make predctions
  output=model(distances)
  # 2.Calculate the loss: how far off was this prediction was?
  loss = loss_function(output, time)
  # 3.calculate adjustments : internal parameters to improve the predictions
  loss.backward() #back propagation
  # 4.update the model (parameters)
  optimizer.step()
```

```python
#Inference

with torch.no_grad(): #not training anymore
  test_distance = torch.tensor([[25.0]], dtype=torch.float32)
  predicted_time = model(test_distance)
  print(f"Predicted time for 25 miles:{predicted_time.item():.1f} minutes")
```
